```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import platform

# 1. 폰트 및 기본 설정
if platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

# 💡 사용자님의 로컬 작업 폴더 경로 설정
base_path = r"C:\Project\Python_Source\python311_01\pandas_ex\project_1"

def load_data(filename):
    file_path = os.path.join(base_path, filename)
    encodings = ['cp949', 'utf-8-sig', 'utf-8', 'euc-kr']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except (UnicodeDecodeError, LookupError):
            continue
        except FileNotFoundError:
            print(f"❌ 파일 없음 에러: 폴더에 '{filename}' 파일이 없습니다.")
            return None
    return None

def clean_route_name(name):
    if pd.isna(name): return "기타"
    name = str(name).strip()
    if "(" in name: name = name.split("(")[0].strip() 
    return name

# =====================================================================
# [Step 1] 데이터 로드 및 위험도 지수 계산
# =====================================================================
print("데이터를 로드하고 위험도를 계산합니다...")
acc_ex = load_data("한국도로공사_졸음운전 사고통계(2022~2024).csv") 
shelters = load_data("한국도로공사 졸음쉼터 설치현황(2024년).csv") 

# 노선명 표준화
acc_ex['정제노선'] = acc_ex['노선명'].apply(clean_route_name)
shelters['정제노선'] = shelters['노선'].apply(clean_route_name)

# 사고수 및 쉼터수 집계
acc_route = acc_ex.groupby('정제노선').size().reset_index(name='사고수')
she_route = shelters.groupby('정제노선').size().reset_index(name='쉼터수')

# 데이터 병합
df_risk = pd.merge(acc_route, she_route, on='정제노선', how='left').fillna(0)

# 💡 핵심 공식: 위험도 지수 = 사고빈도 / 쉼터수
# (단, 쉼터가 0개인 노선은 분모가 0이 되어 에러가 나므로, 수학적 계산을 위해 임시로 분모를 1로 처리)
df_risk['계산용_쉼터수'] = df_risk['쉼터수'].replace(0, 1)
df_risk['위험도지수'] = df_risk['사고수'] / df_risk['계산용_쉼터수']

# 고위험군 Top 5 선정
top5_risk = df_risk.sort_values('위험도지수', ascending=False).head(5)

# =====================================================================
# 시각화: 고위험군 Top 5
# =====================================================================
plt.figure(figsize=(10, 6))
sns.barplot(data=top5_risk, x='정제노선', y='위험도지수', hue='정제노선', palette='Reds_r', legend=False)

plt.title('[Step 1] 노선별 졸음사고 위험도 지수 Top 5 (사고빈도 ÷ 쉼터수)', fontsize=15, fontweight='bold', pad=20)
plt.xlabel('고속도로 노선명', fontsize=12)
plt.ylabel('위험도 지수', fontsize=12)

# 수치 표시 및 쉼터 개수 표기
for i, row in enumerate(top5_risk.itertuples()):
    # 막대 위에 위험도 수치와 실제 쉼터 개수를 같이 적어줌
    text_label = f"위험도: {row.위험도지수:.1f}\n(사고 {row.사고수}건 / 쉼터 {int(row.쉼터수)}개)"
    plt.text(i, row.위험도지수 + 0.1, text_label, ha='center', fontweight='bold')

plt.ylim(0, top5_risk['위험도지수'].max() * 1.2) # 글자가 잘리지 않게 높이 여백 추가
plt.tight_layout()
plt.show()

print("\n🎯 [고위험군 선정 결과]")
print(top5_risk[['정제노선', '사고수', '쉼터수', '위험도지수']].to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import platform

# 1. 폰트 및 기본 설정
if platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

base_path = r"C:\Project\Python_Source\python311_01\pandas_ex\project_1"

def load_data(filename):
    file_path = os.path.join(base_path, filename)
    encodings = ['cp949', 'utf-8-sig', 'utf-8', 'euc-kr']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except (UnicodeDecodeError, LookupError):
            continue
        except FileNotFoundError:
            print(f"❌ 파일 없음 에러: '{filename}' 파일이 없습니다.")
            return None
    return None

def clean_route_name(name):
    if pd.isna(name): return "기타"
    name = str(name).strip()
    if "(" in name: name = name.split("(")[0].strip() 
    return name

# =====================================================================
# [Step 1] 데이터 로드 및 정밀 위험도 계산
# =====================================================================
print("데이터를 로드하고 정밀 위험도를 계산합니다...")
acc_ex = load_data("한국도로공사_졸음운전 사고통계(2022~2024).csv") 
shelters = load_data("한국도로공사 졸음쉼터 설치현황(2024년).csv") 
traffic_main = load_data("한국도로공사_노선별 교통량_20241231 (1).csv") 

# 노선명 표준화
acc_ex['정제노선'] = acc_ex['노선명'].apply(clean_route_name)
shelters['정제노선'] = shelters['노선'].apply(clean_route_name)
traffic_main['정제노선'] = traffic_main['구분'].apply(clean_route_name)

# 1. 사고수 및 쉼터수 집계
acc_route = acc_ex.groupby('정제노선').size().reset_index(name='사고수')
she_route = shelters.groupby('정제노선').size().reset_index(name='쉼터수')

# 2. 본선 교통량 집계 (최근 5년 평균)
traffic_main['5년평균교통량'] = traffic_main[['2020년', '2021년', '2022년', '2023년', '2024년']].mean(axis=1)
tr_main_route = traffic_main.groupby('정제노선')['5년평균교통량'].sum().reset_index(name='본선교통량(천대)')
tr_main_route['본선교통량(천대)'] = tr_main_route['본선교통량(천대)'] / 1000

# 3. 데이터 병합
df_risk = pd.merge(acc_route, she_route, on='정제노선', how='outer').fillna(0)
df_risk = pd.merge(df_risk, tr_main_route, on='정제노선', how='inner')

# 💡 [핵심] 쉼터가 0개인 곳과 교통량 정보가 없는 곳은 완전히 제외!
df_risk = df_risk[(df_risk['쉼터수'] > 0) & (df_risk['본선교통량(천대)'] > 0)].copy()

# 4. 신빙성 높은 최종 위험도 지수 산출
# 공식: (사고수 / 교통량) / 쉼터수 * 스케일링(1000)
df_risk['위험도지수'] = (df_risk['사고수'] / df_risk['본선교통량(천대)']) / df_risk['쉼터수'] * 1000

# 고위험군 Top 5 선정
top5_risk = df_risk.sort_values('위험도지수', ascending=False).head(5)

# =====================================================================
# 시각화: 개선된 고위험군 Top 5
# =====================================================================
plt.figure(figsize=(12, 6))
sns.barplot(data=top5_risk, x='정제노선', y='위험도지수', hue='정제노선', palette='OrRd_r', legend=False)

plt.title('[Step 1] 통행량 반영 정밀 위험도 Top 5 (쉼터 0개 노선 제외)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('고속도로 노선명', fontsize=12)
plt.ylabel('최종 위험도 지수', fontsize=12)

# 수치 표시
for i, row in enumerate(top5_risk.itertuples()):
    text_label = f"위험도: {row.위험도지수:.2f}\n(사고 {int(row.사고수)}건/쉼터 {int(row.쉼터수)}개)"
    plt.text(i, row.위험도지수 + (top5_risk['위험도지수'].max() * 0.02), text_label, ha='center', fontweight='bold')

plt.ylim(0, top5_risk['위험도지수'].max() * 1.2) 
plt.tight_layout()
plt.show()

print("\n🎯 [정밀 필터링 기반 고위험군 선정 결과]")
print(top5_risk[['정제노선', '사고수', '본선교통량(천대)', '쉼터수', '위험도지수']].to_string(index=False, float_format="%.2f"))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import platform
from scipy.stats import gaussian_kde
import warnings

warnings.filterwarnings('ignore')

# =====================================================================
# 1. 폰트 및 기본 설정
# =====================================================================
if platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

# 💡 사용자님의 로컬 작업 폴더 경로 설정
base_path = r"C:\Project\Python_Source\python311_01\pandas_ex\project_1"

def load_data(filename):
    file_path = os.path.join(base_path, filename)
    encodings = ['cp949', 'utf-8-sig', 'utf-8', 'euc-kr']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except (UnicodeDecodeError, LookupError):
            continue
        except FileNotFoundError:
            print(f"❌ 파일 없음 에러: '{filename}' 파일이 없습니다.")
            return None
    return None

def clean_route_name(name):
    if pd.isna(name): return "기타"
    name = str(name).strip()
    if "(" in name: name = name.split("(")[0].strip() 
    return name

# =====================================================================
# 2. 데이터 로드 및 전처리
# =====================================================================
print("데이터를 로드하고 공간 분석을 준비합니다...")
acc_ex = load_data("한국도로공사_졸음운전 사고통계(2022~2024).csv") 
shelters = load_data("한국도로공사 졸음쉼터 설치현황(2024년).csv") 
traffic_main = load_data("한국도로공사_노선별 교통량_20241231 (1).csv") 

acc_ex['정제노선'] = acc_ex['노선명'].apply(clean_route_name)
shelters['정제노선'] = shelters['노선'].apply(clean_route_name)
traffic_main['정제노선'] = traffic_main['구분'].apply(clean_route_name)

# 💡 이정(km) 데이터 숫자 변환
acc_ex['사고발생이정'] = pd.to_numeric(acc_ex['사고발생이정'], errors='coerce')
shelters['이정'] = pd.to_numeric(shelters['이정'], errors='coerce')

# =====================================================================
# 3. [Step 1 복습] Top 5 위험 노선 자동 추출
# =====================================================================
acc_route = acc_ex.groupby('정제노선').size().reset_index(name='사고수')
she_route = shelters.groupby('정제노선').size().reset_index(name='쉼터수')

traffic_main['5년평균교통량'] = traffic_main[['2020년', '2021년', '2022년', '2023년', '2024년']].mean(axis=1)
tr_main_route = traffic_main.groupby('정제노선')['5년평균교통량'].sum().reset_index(name='본선교통량(천대)')
tr_main_route['본선교통량(천대)'] = tr_main_route['본선교통량(천대)'] / 1000

df_risk = pd.merge(acc_route, she_route, on='정제노선', how='outer').fillna(0)
df_risk = pd.merge(df_risk, tr_main_route, on='정제노선', how='inner')
df_risk = df_risk[(df_risk['쉼터수'] > 0) & (df_risk['본선교통량(천대)'] > 0)].copy()

df_risk['위험도지수'] = (df_risk['사고수'] / df_risk['본선교통량(천대)']) / df_risk['쉼터수'] * 1000
# 💡 대상을 Top 5로 확대
top5_routes = df_risk.sort_values('위험도지수', ascending=False).head(5)['정제노선'].tolist()

# =====================================================================
# 4. [Step 2] Top 5 노선별 입지 최적화 (Kernel Density Estimation)
# =====================================================================
# 💡 5개의 그래프를 그리기 위해 세로 길이를 20으로 여유있게 늘림
fig, axes = plt.subplots(5, 1, figsize=(14, 20), sharex=False)
plt.suptitle('[Step 2] 고위험 노선 Top 5: 사고 밀집 구역 및 최적 쉼터 설치 제안 지점', fontsize=20, fontweight='bold', y=0.97)

for i, route in enumerate(top5_routes):
    ax = axes[i]
    
    # 해당 노선의 사고 및 쉼터 데이터 필터링
    r_acc = acc_ex[(acc_ex['정제노선'] == route)].dropna(subset=['사고발생이정'])
    r_she = shelters[shelters['정제노선'] == route].dropna(subset=['이정'])
    
    if len(r_acc) > 1:
        # ① 사고 밀도 시각화 (KDE)
        sns.kdeplot(data=r_acc, x='사고발생이정', fill=True, color='red', alpha=0.2, ax=ax, label='사고 발생 밀도 (위험구간)')
        
        # ② 수학적 극대값(가장 사고가 많이 몰린 지점) 도출 알고리즘
        kde = gaussian_kde(r_acc['사고발생이정'])
        x_range = np.linspace(r_acc['사고발생이정'].min(), r_acc['사고발생이정'].max(), 1000)
        kde_vals = kde(x_range)
        peak_idx = np.argmax(kde_vals)
        peak_loc = x_range[peak_idx] # 최적 제안 지점
        
        # ③ 실제 사고 발생 위치 점 찍기
        ax.scatter(r_acc['사고발생이정'], [0]*len(r_acc), color='darkred', marker='x', s=30, label='개별 사고 발생 지점')
        
        # ④ 현재 쉼터 위치 표시
        for loc in r_she['이정']:
            ax.axvline(loc, color='blue', linestyle='--', linewidth=2.5, alpha=0.6)
        
        # 범례용 가짜 선 (파란 점선)
        if len(r_she) > 0:
            ax.axvline(-1000, color='blue', linestyle='--', linewidth=2.5, alpha=0.6, label='현재 운영 중인 쉼터 위치') 
            
        # ⑤ 💡 최적 설치 제안 지점 표시 (초록색 굵은 선)
        ax.axvline(peak_loc, color='green', linestyle='-', linewidth=4, label=f'⭐ 1순위 설치 제안 지점 ({peak_loc:.1f}km)')
        
        ax.set_title(f'Top {i+1}. {route} (위험도 {i+1}순위 타겟)', fontsize=15, fontweight='bold', pad=10)
        ax.set_xlim(max(0, r_acc['사고발생이정'].min()-10), r_acc['사고발생이정'].max()+10)
        ax.set_xlabel('도로 이정 (km)', fontsize=12)
        ax.set_ylabel('사고 밀집도', fontsize=12)
        ax.set_yticks([]) 
        ax.legend(loc='upper right', frameon=True, shadow=True, fontsize=10)
    else:
        ax.set_title(f'Top {i+1}. {route} - 사고 위치 데이터 부족으로 분석 불가', fontsize=15, fontweight='bold')
        ax.set_yticks([])

plt.tight_layout()
plt.subplots_adjust(top=0.93) # 메인 타이틀 공간 확보
plt.show()

print("\n🎯 [데이터 기반 입지 선정 분석 완료: Top 5 노선]")
```
