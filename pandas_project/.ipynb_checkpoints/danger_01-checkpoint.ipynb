```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 1. 시각화 한글 설정
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 작업 경로 지정 (사용자 환경)
base_path = r"C:\Project\Python_Source\python311_01\pandas_ex\project_1"

# 3. 경로가 반영된 파일 로더 함수
def read_korean_csv(filename):
    file_path = os.path.join(base_path, filename)
    encodings = ['cp949', 'utf-8-sig', 'utf-8', 'euc-kr']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except (UnicodeDecodeError, LookupError):
            continue
        except FileNotFoundError:
            print(f"❌ 오류: '{filename}' 파일을 찾을 수 없습니다. 폴더에 파일이 있는지 확인해주세요.")
            return None
    return None

# 파일 로드
df_acc_route = read_korean_csv('한국도로공사_졸음운전 사고통계(2022~2024).csv')
df_shelter = read_korean_csv('한국도로공사 졸음쉼터 설치현황(2024년).csv')

# 파일이 정상적으로 로드되었는지 체크
if df_acc_route is None or df_shelter is None:
    raise ValueError("파일 로딩에 실패하여 분석을 중단합니다. 위 에러 메시지를 확인해주세요.")

# ==========================================
# 1단계: 노선명 표준화
# ==========================================
def clean_route_name(name):
    if pd.isna(name): return "기타"
    name = str(name).strip()
    if "(" in name:
        name = name.split("(")[0]
    return name

df_acc_route['정제노선'] = df_acc_route['노선명'].apply(clean_route_name)
df_shelter['정제노선'] = df_shelter['노선'].apply(clean_route_name)

# ==========================================
# 2단계: 노선별 사고 건수 및 쉼터 개수 집계
# ==========================================
acc_counts = df_acc_route.groupby('정제노선').size().reset_index(name='사고건수')
shelter_counts = df_shelter.groupby('정제노선').size().reset_index(name='쉼터수')

risk_df = pd.merge(acc_counts, shelter_counts, on='정제노선', how='outer').fillna(0)

# ==========================================
# 3단계: 위험도 지수 산출 (사고건수 / (쉼터수 + 1))
# ==========================================
# 쉼터가 0개인 곳의 분모 0 에러 방지를 위해 분모에 +1 처리
risk_df['위험도지수'] = risk_df['사고건수'] / (risk_df['쉼터수'] + 1)

# 위험도 순으로 정렬 (TOP 10 추출)
risk_top10 = risk_df.sort_values(by='위험도지수', ascending=False).head(10)

# ==========================================
# 4단계: 시각화 및 결과 출력
# ==========================================
plt.figure(figsize=(12, 7))
sns.barplot(data=risk_top10, x='정제노선', y='위험도지수', hue='쉼터수', palette='OrRd_r', dodge=False)

plt.title('고속도로 노선별 위험도 지수 (사고 빈도 대비 쉼터 부족 현황)', fontsize=16, pad=20, fontweight='bold')
plt.xlabel('노선명', fontsize=12)
plt.ylabel('위험도 지수 (사고건수 / (현재 쉼터 수 + 1))', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='현재 쉼터 수 (개)')

# 수치 텍스트 표시
for i, v in enumerate(risk_top10['위험도지수']):
    plt.text(i, v + 0.2, f"{v:.1f}", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()

# 데이터프레임 텍스트 출력
print("\n📊 [노선별 위험도 지수 TOP 10 (우선 설치 타겟)]")
print("-" * 55)
print(risk_top10[['정제노선', '사고건수', '쉼터수', '위험도지수']].to_string(index=False, float_format='%.1f'))
print("-" * 55)

# CSV 저장
save_path = os.path.join(base_path, "노선별_위험도_분석_결과.csv")
risk_top10.to_csv(save_path, index=False, encoding='utf-8-sig')
```
