```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import platform
import warnings

warnings.filterwarnings('ignore')

# =====================================================================
# 1. 폰트 및 기본 설정
# =====================================================================
if platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

base_path = r"C:\Project\Python_Source\python311_01\pandas_ex\project_1"

def load_data(filename):
    file_path = os.path.join(base_path, filename)
    encodings = ['utf-8-sig', 'cp949', 'utf-8', 'euc-kr']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except:
            continue
    print(f"❌ 파일 로드 실패: {filename}")
    return None

def clean_route_name(name):
    if pd.isna(name): return "기타"
    name = str(name).strip()
    if "(" in name: name = name.split("(")[0].strip() 
    return name

# =====================================================================
# 2. 데이터 로드 및 전처리
# =====================================================================
print("2중 필터링 기준을 적용하여 타겟 노선을 추출합니다...")
acc_ex = load_data("한국도로공사_졸음운전 사고통계(2022~2024).csv") 
shelters = load_data("한국도로공사 졸음쉼터 설치현황(2024년).csv") 
traffic_main = load_data("한국도로공사_노선별 교통량_20241231 (1).csv") 

acc_ex['정제노선'] = acc_ex['노선명'].apply(clean_route_name)
shelters['정제노선'] = shelters['노선'].apply(clean_route_name)
traffic_main['정제노선'] = traffic_main['구분'].apply(clean_route_name)

# 집계 (사고수, 쉼터수, 본선교통량)
acc_route = acc_ex.groupby('정제노선').size().reset_index(name='사고수')
she_route = shelters.groupby('정제노선').size().reset_index(name='쉼터수')

traffic_main['5년평균교통량'] = traffic_main[['2020년', '2021년', '2022년', '2023년', '2024년']].mean(axis=1)
tr_main_route = traffic_main.groupby('정제노선')['5년평균교통량'].sum().reset_index(name='본선교통량(천대)')
tr_main_route['본선교통량(천대)'] = tr_main_route['본선교통량(천대)'] / 1000

# 데이터 병합
df_risk = pd.merge(acc_route, she_route, on='정제노선', how='outer').fillna(0)
df_risk = pd.merge(df_risk, tr_main_route, on='정제노선', how='inner')

# =====================================================================
# 3. 💡 [핵심] 우리가 세운 '비즈니스 로직' 필터링 적용
# =====================================================================
# 필터 1: 쉼터 0개 및 교통량 없는 곳 제외 (기본 조건)
df_risk = df_risk[(df_risk['쉼터수'] > 0) & (df_risk['본선교통량(천대)'] > 0)]

# 필터 2: 누적 사고 10건 미만 배제 (비율 왜곡/통계적 노이즈 제거)
min_accidents = 10
df_risk_filtered = df_risk[df_risk['사고수'] >= min_accidents].copy()

# 위험도 산출
df_risk_filtered['위험도지수'] = (df_risk_filtered['사고수'] / df_risk_filtered['본선교통량(천대)']) / df_risk_filtered['쉼터수'] * 1000

# 필터 3: 현실적 예산 제약을 고려한 'Phase 1' 최상위 5개소 추출
top5_risk = df_risk_filtered.sort_values('위험도지수', ascending=False).head(5)

# =====================================================================
# 4. 시각화
# =====================================================================
plt.figure(figsize=(12, 7))

sns.barplot(data=top5_risk, x='정제노선', y='위험도지수', hue='정제노선', palette='OrRd_r', legend=False)

# 차트 타이틀 및 부제목 (우리의 논리를 명시)
plt.title('졸음쉼터 신설 Phase 1 타겟: 2중 필터링 검증 고위험군 Top 5', fontsize=17, fontweight='bold', pad=30)
plt.suptitle('※ 선정 기준: ① 3년 누적 사고 10건 이상 (통계 착시 배제)  ② 예산/시공 여력 고려 상위 5개소 한정', 
             fontsize=12, color='gray', y=0.92)

plt.xlabel('고속도로 노선명', fontsize=13, fontweight='bold')
plt.ylabel('최종 위험도 지수', fontsize=13, fontweight='bold')

# 막대 위에 수치 및 상세 스펙 텍스트 표시
for i, row in enumerate(top5_risk.itertuples()):
    text_label = f"위험도: {row.위험도지수:.2f}\n(사고 {int(row.사고수)}건/쉼터 {int(row.쉼터수)}개)"
    plt.text(i, row.위험도지수 + (top5_risk['위험도지수'].max() * 0.02), 
             text_label, ha='center', fontweight='bold', fontsize=11, color='#c0392b')

sns.despine(left=False, bottom=False)
plt.ylim(0, top5_risk['위험도지수'].max() * 1.25) 
plt.tight_layout()
plt.show()

# 데이터 프레임 출력
print("\n🎯 [2중 필터링 적용] 전략적 타겟 노선 Top 5 도출 완료")
print("-" * 80)
print(top5_risk[['정제노선', '사고수', '본선교통량(천대)', '쉼터수', '위험도지수']].to_string(index=False, float_format="%.2f"))
print("-" * 80)
```
