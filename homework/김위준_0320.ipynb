### (1) 팁 데이터에서 pivot_table()함수를 사용하여 index='day', columns='size', values='tip'으로하는 데이터를 만든 다음 해당 데이터로 히트맵을 작성하시오.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
tips.head(2)
```

```python
tips.groupby(['total_bill','size'])['tip'].mean()
```

```python
tips1 = tips.pivot_table(index='day', columns = 'size',values = 'tip')
tips1
```

```python
tips1.mean()
```

```python
tips1.fillna(tips1.mean())
```

```python
sns.heatmap(tips1, cmap=sns.light_palette("skyblue", as_cmap=True), annot=True)
plt.title("Tips Heatmap")
plt.show()
```

### (2) 아이리스 데이터에서 각 데이터의 상관관계를 pairplot()으로 나타내시오.

```python
import seaborn as sns


iris = sns.load_dataset("iris")   
plt.rcParams["font.family"] = 'Malgun Gothic'
sns.pairplot(iris, diag_kind="kde", markers= 's', height=3)
plt.title("Iris Data의 Pair Plot")
plt.show()
```
