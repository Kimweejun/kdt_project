### Countplot

```python
import seaborn as sns

titanic = sns.load_dataset("titanic")
sns.countplot(x="class",data=titanic, hue = "who")
plt.rcParams["font.family"] = 'Malgun Gothic'
plt.title("타이타닉호의 각 클래스별, 승객 수")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")     # titanic, tips, flight
x= iris.petal_length.values
plt.rcParams["font.family"] = 'Malgun Gothic'
sns.distplot(x,kde= True, rug=True)
plt.title("iris 데이터 중, 꽃잎의 길이에 대한 Dist Plot")
plt.show
```

```python
import seaborn as sns
import matplotlib as mpl

mpl.rcParams['axes.unicode_minus'] = False
titanic = sns.load_dataset("titanic")
data = titanic[titanic.survived.notnull()]
sns.catplot(x="age", y='sex',hue='survived', row = 'class',data=data,
           kind='violin', split=True, height=2,aspect=4) # aspect 가로세로 비율
plt.title("각 클래스의 성별 생존자 수의 Catplot")
plt.show()
```

```python
import seaborn as sns
import matplotlib as mpl

mpl.rcParams['axes.unicode_minus'] = False
titanic = sns.load_dataset("titanic")
data = titanic[titanic.survived.notnull()]
sns.catplot(x="age", y='sex',hue='survived', row = 'class',data=data,
           kind='bar', height=2,aspect=4) 
plt.suptitle("각 클래스의 성별 생존자 수의 Catplot")
plt.show()
```

```python
sns.catplot(x='who', y='survived', col='pclass',
           row='embarked',
           kind='bar',
           height=2.5,
           data=titanic)
plt.show
```

```python
import seaborn as sns
import numpy as np

tips = sns.load_dataset('tips')
plt.title("Boxplot과 Strip Plot로 표현한 요일 별 팁")
sns.boxplot(x='tip', y='day', data=tips, whis=np.inf)
sns.stripplot(x='tip', y='day', data=tips, jitter=True, color='0.4')
plt.show
```
