```python
import pymysql ## pip install pymysql

conn = pymysql.connect(host='localhost',
                       port=3306,
                       user='root',
                       password='123456',
                       db = 'project',
                       charset='utf8')
cursor = conn.cursor()

sql = "SELECT * FROM 시군구별전력사용량_202512 WHERE `MyUnknownColumn` = '경상북도' AND `MyUnknownColumn_[0]` = '포항시' AND `MyUnknownColumn_[1]` = '합계';"

cursor.execute(sql)

rows = cursor.fetchall()

for row in rows:
    print(row)


cursor.close()
conn.close()

```
