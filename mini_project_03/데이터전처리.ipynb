```python
import pandas as pd

file_path = '포항시남구표제부1.csv'

try:
    df = pd.read_csv(file_path, encoding='cp949', dtype={'법정동코드': str, '번': str, '지': str})
except UnicodeDecodeError:
    df = pd.read_csv(file_path, encoding='utf-8', dtype={'법정동코드': str, '번': str, '지': str})

df['법정동코드'] = df['법정동코드'].fillna('')
df['번'] = df['번'].fillna('')
df['지'] = df['지'].fillna('')

df['관리번호_PK'] = df['법정동코드'] + '-' + df['번'] + '-' + df['지']

# 중복 데이터 제거! ★ (이 부분이 추가되었습니다)
# subset=['관리번호_PK']: 이 컬럼을 기준으로 중복 검사
# keep='first': 중복된 행 중 맨 처음에 나온 데이터만 남기고 삭제
original_count = len(df)
df = df.drop_duplicates(subset=['관리번호_PK'], keep='first')
new_count = len(df)
print(f"✔️ 중복 제거 완료: 총 {original_count}개 중 {original_count - new_count}개의 중복 데이터가 삭제되어 {new_count}개만 남았습니다.\n")

df = df.drop(columns=['법정동코드', '번', '지'])

cols = ['관리번호_PK'] + [col for col in df.columns if col != '관리번호_PK']
df = df[cols]

# 결과 확인
print("기존 컬럼이 지워지고 PK가 맨 앞에 잘 배치되었는지 확인해 보세요:")
display(df.head())

output_path = '포항남구표제부_PK추가_clean.csv'
df.to_csv(output_path, encoding='utf-8-sig', index=False)

print(f"\n변환 완료! '{output_path}' 파일이 생성되었습니다.")
```
