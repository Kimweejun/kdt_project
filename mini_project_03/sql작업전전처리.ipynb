```python
import pandas as pd

# 1. 데이터 불러오기
try:
    df = pd.read_csv('포항시전력사용량20_22.csv', encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv('포항시전력사용량20_22.csv', encoding='cp949')

data = df.iloc[4:].copy()
data['월'] = data['월별(1)'].str.replace('월', '').astype(int)

df
```

```python
# 2. 연도별/용도별 컬럼 매핑
years = ['2020', '2021', '2022']
col_mapping = {
    '2019': {'가정용': '2019', '공공용': '2019.2', '서비스업': '2019.4', '산업용': '2019.6'},
    '2020': {'가정용': '2020', '공공용': '2020.2', '서비스업': '2020.4', '산업용': '2020.6'},
    '2021': {'가정용': '2021', '공공용': '2021.2', '서비스업': '2021.4', '산업용': '2021.6'},
    '2022': {'가정용': '2022', '공공용': '2022.2', '서비스업': '2022.4', '산업용': '2022.6'},
    '2023': {'가정용': '2023', '공공용': '2023.2', '서비스업': '2023.4', '산업용': '2023.6'}
}

# 3. SQL DB 형태(Long Form)로 데이터 변환
rows = []
for year in years:
    for _, row in data.iterrows():
        month = row['월']
        home = int(pd.to_numeric(row[col_mapping[year]['가정용']], errors='coerce'))
        public = int(pd.to_numeric(row[col_mapping[year]['공공용']], errors='coerce'))
        service = int(pd.to_numeric(row[col_mapping[year]['서비스업']], errors='coerce'))
        industry = int(pd.to_numeric(row[col_mapping[year]['산업용']], errors='coerce'))
        total = home + public + service + industry
        
        rows.append({
            '연도': int(year),
            '월': month,
            '가정용_사용량': home,
            '공공용_사용량': public,
            '서비스업_사용량': service,
            '산업용_사용량': industry,
            '총사용량': total
        })

# 4. 파일 저장
clean_df = pd.DataFrame(rows)
clean_df.to_csv('포항시전력사용량_DB용.csv', index=False, encoding='utf-8-sig')
print("DB 임포트용 파일이 생성되었습니다!")

clean_df
```

```python
import pandas as pd

# 1. 원본 파일 불러오기
try:
    df = pd.read_csv('포항시전력사용량11_20.csv', encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv('포항시전력사용량11_20.csv', encoding='cp949')

# 2. 동적 컬럼 매핑 ('소계' 대신 '사용량'이라는 단어 기준)
years = list(range(2011, 2021))
sectors = ['가정용', '공공용', '서비스업 및 기타', '산업용', '합계'] # 총사용량을 추출하기 위해 '합계' 추가
mapping = {str(y): {} for y in years}

for col in df.columns:
    if col == '월별(1)': continue
    year_str = col.split('.')[0]
    
    if not year_str.isdigit(): continue
    if int(year_str) not in years: continue
    
    # 엑셀의 다중 헤더 텍스트 추출
    val0 = str(df.loc[0, col]).strip() # 가정용, 공공용, 합계 등
    val2 = str(df.loc[2, col]).strip() # 사용량 (MWh), 금액 등
    
    # ★ 핵심 변경: val1('소계') 조건을 빼고, val2에 '사용량'이 들어간 컬럼만 정확히 타겟팅
    if '사용량' in val2:
        for sec in sectors:
            # 중복 방지를 위해 아직 해당 연도/용도의 컬럼이 지정되지 않았을 때만 저장
            if val0 == sec and sec not in mapping[year_str]:
                mapping[year_str][sec] = col

# ★ 추가된 계절 판별 함수
def get_season(month):
    if month in [3, 4, 5]:
        return '봄'
    elif month in [6, 7, 8]:
        return '여름'
    elif month in [9, 10, 11]:
        return '가을'
    else:
        return '겨울'

# 3. SQL DB 형태(Long Form)로 데이터 변환
rows = []
for year in [str(y) for y in years]:
    for m in range(1, 13): # 1월 ~ 12월
        row_idx = m + 4    # CSV 파일 구조상 1월 데이터는 인덱스 5부터 시작
        
        try:
            # 콤마(,) 제거 후 숫자로 변환
            home = int(float(str(df.loc[row_idx, mapping[year]['가정용']]).replace(',', '')))
            public = int(float(str(df.loc[row_idx, mapping[year]['공공용']]).replace(',', '')))
            service = int(float(str(df.loc[row_idx, mapping[year]['서비스업 및 기타']]).replace(',', '')))
            industry = int(float(str(df.loc[row_idx, mapping[year]['산업용']]).replace(',', '')))
            
            # 총사용량 역시 수동 계산이 아닌 원본 데이터 '합계'에서 직접 가져옴
            total = int(float(str(df.loc[row_idx, mapping[year]['합계']]).replace(',', '')))
            
            rows.append({
                '연도': int(year),
                '월': m,
                '계절': get_season(m), # ★ 계절 컬럼 추가
                '가정용사용량': home,
                '공공용사용량': public,
                '서비스업사용량': service,
                '산업용사용량': industry,
                '총사용량': total
            })
        except (ValueError, KeyError):
            # 데이터가 빈칸('-')이거나 없는 월은 건너뜀
            pass

# 4. 파일 저장 및 확인
clean_df = pd.DataFrame(rows)

# 새로 생성될 파일 이름
clean_df.to_csv('포항시전력사용량_사용량기준_DB용.csv', index=False, encoding='utf-8-sig')

print(f"✅ DB 임포트용 파일이 성공적으로 생성되었습니다! (총 {len(clean_df)}건)")
display(clean_df.head(10)) # 주피터 노트북 출력용
```
