## 표준화

```python
import seaborn as sns

iris = sns.load_dataset("iris")
iris.head(2)
```

```python
x = iris.iloc[:,:-1]
x.head(2)
```

```python
from sklearn.preprocessing import scale

x_scaled = scale(x)
x_scaled[:5,:]
```

```python
x_scaled.mean(axis=0)
```

```python
for scaled_mean in x_scaled.mean(axis=0):
    print("{:10.9f}".format(scaled_mean))
```

```python
from sklearn.preprocessing import robust_scale

x_robust_scaled = robust_scale(x)
x_robust_scaled[:5,:]
```

```python
from sklearn.preprocessing import minmax_scale

x_minmax_scaled = minmax_scale(x)
x_minmax_scaled[:5,:]
```

```python
from sklearn.preprocessing import maxabs_scale

x_maxabs_scaled = maxabs_scale(x)
x_maxabs_scaled[:5,:]
```

```python
import seaborn as sns
iris = sns.load_dataset("iris")
x=iris.iloc[:,:-1]
x
```

```python
import random

for col in range(4):
    x.iloc[[random.sample(range(len(iris)),10)],col] = float('nan')
x.head()
```

```python
x.mean(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
imp_mean.fit_transform(x)[0:5,:]
```

```python
x.median(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_median = SimpleImputer(missing_values=np.nan, strategy='median')
imp_median.fit_transform(x)[0:5,:]
```

```python
x.mode(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_most_frequent = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
imp_most_frequent.fit_transform(x)[0:5,:]
```
