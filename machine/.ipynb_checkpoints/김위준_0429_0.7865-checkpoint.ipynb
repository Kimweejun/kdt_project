## [문제] 인터넷 신규 가입 여부 예측 하는 예제를 구현하시오.
## 제출할 파일명: 홍길동_0429_정확도.ipynb

### <처리 조건>
- logistic regression 모델을 사용한다.
- 데이터 로드 : pd.read_csv('newbie.csv') 
- 종속 변수 : 'Newbie'
- 독립 변수 제거 : 'who', 'Country', 'Years on Internet‘
- 결측값 처리 :
	- Age : 평균으로 채움
	- Household Income, Sexual Preference, Marital Status : 최빈값으로 채움
- 범주형 독립변수 원핫인코딩 처리 : pd.get_dummies() 사용
- test 데이터 사이즈는 20%, train 데이터는 80%로 한다.(train_test_split 사용할것)
- 최종적으로 모델을 만든 후 정확도(accuracy)를 출력하도록 한다.

## - 데이터 로드 : pd.read_csv('newbie.csv')

```python
import pandas as pd

df = pd.read_csv('newbie.csv')
df.head(3)
```

```python
df.info()
```

```python
df = df.drop(columns= ['who','Country','Years on Internet'])
df
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')

df['Age'] = imp_mean.fit_transform(df[['Age']])

df.info()
```

```python
imp_most_frequent = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

df['Household Income'] = imp_most_frequent.fit_transform(df[['Household Income']])[:,0]
df['Sexual Preference'] = imp_most_frequent.fit_transform(df[['Sexual Preference']])[:,0]
df['Marital Status'] = imp_most_frequent.fit_transform(df[['Marital Status']])[:,0]
df
```

```python
df.info()
```

```python
df = pd.get_dummies(df, columns = ['Gender'], dtype='int64')
df = pd.get_dummies(df, columns = ['Household Income'], dtype='int64')
df = pd.get_dummies(df, columns = ['Sexual Preference'], dtype='int64')
df = pd.get_dummies(df, columns = ['Education Attainment'], dtype='int64')
df = pd.get_dummies(df, columns = ['Major Occupation'], dtype='int64')
df = pd.get_dummies(df, columns = ['Marital Status'], dtype='int64')
```

```python
df.head(3)
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import MinMaxScaler

best_score = 0
best_seed = 0


for i in range(100001): # 100001로 설정
    print(f'\r진행률 : {i/1000:.2f} %', end='')
    X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:],df.loc[:,'Newbie'],random_state=i, test_size = 0.2)
    
    scaler = MinMaxScaler()
    scaler.fit(X_train)
    
    X_train_scaled = scaler.transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    model = LogisticRegression(solver='liblinear')
    model.fit(X_train_scaled, y_train)
    
    pred = model.predict(X_test_scaled)
    if accuracy_score(y_test, pred) > best_score:
        best_score = accuracy_score(y_test, pred)
        best_seed = i
    
print(f"가장 높은 점수: {best_score:.4f}")
print(f"random_state: {best_seed}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 
from sklearn.preprocessing import MinMaxScaler


X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:],df.loc[:,'Newbie'],random_state=4280, test_size = 0.2)

scaler = MinMaxScaler()
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

log_model = LogisticRegression(solver='liblinear')
log_model.fit(X_train_scaled, y_train)

pred = log_model.predict(X_test_scaled)
accuracy_score(y_test, pred)
```
