```python
import pandas as pd

df = pd.read_csv("train.csv", encoding='utf-8')
df.head(3)
```

```python
df.info()
```

### (1) 결측값 처리: 처리한 결과를 기존 데이터 프레임에 재저장
- Age : 평균값으로 결측값 처리
- Embarked : 최빈값으로 결측값 처리

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')

df['Age'] = imp_mean.fit_transform(df[['Age']])

df.head(3)
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode())
```

### (2) 레이블 인코딩: 기존 데이터 프레임을 새로은 데이터 프레임으로 저장한 다음 처리한 결과를 기존 데이터 컬럼은 삭제하고 새로운 데이터 프레임에 재저장

```python
df1 = df.copy()
```

```python
from sklearn.preprocessing import LabelEncoder

y1 = df1.Sex
y2 = df1.Embarked

le = LabelEncoder()

df1['Sex'] = le.fit_transform(y1)
df1['Embarked'] = le.fit_transform(y2)
```

```python
df.head(3)
```

### (3) 원핫인코딩: 기존 데이터 프레임을 새로은 데이터 프레임으로 저장한 다음 처리한 결과를 기존 데이터 컬럼은 삭제하고 원핫인코더로 변환한 내용을 새로운 데이터 프레임에 추가할때 컬럼명은 적당히 정해서 추가한다.

```python
df2 = df.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder

enc1 = OneHotEncoder()
enc2 = OneHotEncoder()

enc1.fit(df[['Sex']])
enc2.fit(df[['Embarked']])

Sex_onehot = enc1.transform(df[['Sex']])
Emb_onehot = enc2.transform(df[['Embarked']])

Sex_onehot = Sex_onehot.toarray()
Emb_onehot = Emb_onehot.toarray()

df2 = df.drop(df2[['Sex']],axis=1)
df2 = df.drop(df2[['Embarked']],axis=1)
```

```python
df2['Sex_Female'] = Sex_onehot[:,0]
df2['Sex_Male'] = Sex_onehot[:,1]

df2['Embarked_C'] = Emb_onehot[:,0]
df2['Embarked_Q'] = Emb_onehot[:,1]
df2['Embarked_S'] = Emb_onehot[:,2]
```
