```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 한글 폰트 설정 (깨짐 방지)
# 윈도우 환경인 경우
plt.rc('font', family='Malgun Gothic') 
# Mac 환경인 경우 아래 코드의 주석을 해제하고 위 코드를 주석 처리하세요.
# plt.rc('font', family='AppleGothic')

# 마이너스 기호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False

# 2. 데이터 불러오기
# 첫 번째 행(인덱스 0)은 연도이므로, 두 번째 행(인덱스 1)을 컬럼 헤더로 사용합니다.
file_path = '스포츠_중계_시청을_위하여_OTT_구독_여부_20260412215110.csv'
df = pd.read_csv(file_path, header=1)

# 3. '전체' 데이터 필터링
# '특성별(1)' 컬럼이 '전체'인 행만 추출합니다.
total_data = df[df['특성별(1)'] == '전체']

# 4. 그래프용 데이터 준비
# '예 (%)'와 '아니오 (%)' 값을 추출합니다.
labels = ['예 (%)', '아니오 (%)']
values = [total_data['예 (%)'].values[0], total_data['아니오 (%)'].values[0]]

# 5. 파이 차트 그리기
plt.figure(figsize=(8, 6))
plt.pie(values, 
        labels=labels, 
        autopct='%1.1f%%', # 소수점 첫째 자리까지 % 표시
        startangle=90,     # 12시 방향부터 시작
        colors=['#66b3ff', '#ff9999'], # 파란색, 붉은색 계열 지정
        explode=(0.05, 0)) # '예' 부분을 살짝 강조

plt.title('스포츠 중계 시청을 위한 OTT 구독 여부 (전체 기준)', fontsize=14, pad=15)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. 한글 폰트 설정 (깨짐 방지)
# 윈도우 환경인 경우
plt.rc('font', family='Malgun Gothic') 
# Mac 환경인 경우 아래 코드의 주석을 해제하고 위 코드를 주석 처리하세요.
# plt.rc('font', family='AppleGothic')

# 마이너스 기호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False

# 2. 데이터 불러오기
file_path = '스포츠_중계_시청을_위하여_OTT_구독_여부_20260412215110.csv'
df = pd.read_csv(file_path, header=1)

# 3. '성별' 데이터 필터링
gender_data = df[df['특성별(1)'] == '성별']

# 4. 그래프용 데이터 준비
labels = gender_data['특성별(2)'].tolist()  # ['남성', '여성']
yes_values = gender_data['예 (%)'].tolist()  # 남성/여성의 '예' 비율
no_values = gender_data['아니오 (%)'].tolist() # 남성/여성의 '아니오' 비율

# 5. 막대그래프 위치 및 두께 설정
x = np.arange(len(labels))  # [0, 1]
width = 0.35  # 막대 두께

# 6. 그래프 그리기
fig, ax = plt.subplots(figsize=(8, 6))

# '예'와 '아니오' 막대를 나란히 배치하기 위해 x 좌표를 살짝 이동시킵니다.
rects1 = ax.bar(x - width/2, yes_values, width, label='예 (%)', color='#66b3ff')
rects2 = ax.bar(x + width/2, no_values, width, label='아니오 (%)', color='#ff9999')

# 축 라벨, 타이틀, 범례 설정
ax.set_ylabel('비율 (%)', fontsize=12)
ax.set_title('스포츠 중계 시청을 위한 OTT 구독 여부 (성별 비교)', fontsize=14, pad=15)
ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=12)
ax.legend()

# 막대 그래프 위에 정확한 수치 텍스트 표시
ax.bar_label(rects1, padding=3, fmt='%.1f%%')
ax.bar_label(rects2, padding=3, fmt='%.1f%%')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. 한글 폰트 및 마이너스 기호 설정 (깨짐 방지)
# 윈도우 환경인 경우
plt.rc('font', family='Malgun Gothic') 
# Mac 환경인 경우 아래 코드의 주석을 해제하고 위 코드를 주석 처리하세요.
# plt.rc('font', family='AppleGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 데이터 불러오기 (두 번째 행을 헤더로 사용)
file_path = '스포츠_중계_시청을_위하여_OTT_구독_여부_20260412215110.csv'
df = pd.read_csv(file_path, header=1)

# 3. '연령' 데이터 필터링
age_data = df[df['특성별(1)'] == '연령']

# 4. 그래프용 데이터 준비
labels = age_data['특성별(2)'].tolist()  # ['10대', '20대', '30대', '40대', '50대', '60대', '70세 이상']
yes_values = age_data['예 (%)'].tolist()
no_values = age_data['아니오 (%)'].tolist()

# 5. 막대 위치 및 두께 설정
x = np.arange(len(labels))  # 연령대 개수만큼의 x좌표 [0, 1, 2, ...]
width = 0.35  # 막대 두께

# 6. 그래프 그리기 (가로 크기를 조금 더 넓게 설정)
fig, ax = plt.subplots(figsize=(12, 6))

# '예'와 '아니오' 막대 배치
rects1 = ax.bar(x - width/2, yes_values, width, label='예 (%)', color='#66b3ff')
rects2 = ax.bar(x + width/2, no_values, width, label='아니오 (%)', color='#ff9999')

# 축 라벨, 타이틀, 범례 설정
ax.set_ylabel('비율 (%)', fontsize=12)
ax.set_xlabel('연령대', fontsize=12)
ax.set_title('스포츠 중계 시청을 위한 OTT 구독 여부 (연령대별 비교)', fontsize=15, pad=15)
ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=11)
ax.legend()

# 막대 그래프 위에 정확한 수치 텍스트 표시
ax.bar_label(rects1, padding=3, fmt='%.1f%%')
ax.bar_label(rects2, padding=3, fmt='%.1f%%')

# 레이아웃 조정 및 출력
plt.tight_layout()
plt.show()
```
