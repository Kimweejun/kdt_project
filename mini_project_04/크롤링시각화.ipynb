```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# 1. 한글 폰트 및 그래프 설정
# Windows 사용자는 'Malgun Gothic', Mac 사용자는 'AppleGothic'을 사용하세요.
plt.rcParams['font.family'] = 'Malgun Gothic' 
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 방지
sns.set_theme(style="whitegrid", font=plt.rcParams['font.family'])

# 2. 데이터 불러오기 및 병합
df_movie = pd.read_csv('국내영화_5년내개봉_인기순위100.csv')

# 3. 텍스트 데이터 전처리 및 빈도수 계산 함수
def get_top_items(series, top_n=15):
    # 결측치 제거 후 쉼표(,)를 기준으로 문자열 분리
    items = series.dropna().str.split(',')
    
    # 2차원 리스트를 1차원으로 평탄화하고 양옆 공백 제거
    # 'Vote tags' 등 불필요한 텍스트가 남아있을 경우를 대비해 replace 처리
    flat_list = [
        item.replace('(Vote tags)', '').strip() 
        for sublist in items 
        for item in sublist 
        if item.strip() != ''
    ]
    
    # 빈도수 계산
    counter = Counter(flat_list)
    return pd.DataFrame(counter.most_common(top_n), columns=['Name', 'Count'])

# 장르와 태그의 Top 15 데이터 추출
top_genres = get_top_items(df_movie['Genres'], 15)
top_tags = get_top_items(df_movie['Tags'], 15)

# 4. 시각화 (그래프 그리기)
fig, axes = plt.subplots(2, 1, figsize=(12, 14))

# [상단 그래프] 장르(Genres) 시각화
sns.barplot(data=top_genres, x='Count', y='Name', ax=axes[0], palette='viridis')
axes[0].set_title('Top 15 영화 장르 (Genres) 빈도수', fontsize=16, fontweight='bold')
axes[0].set_xlabel('등장 횟수', fontsize=12)
axes[0].set_ylabel('장르', fontsize=12)

# [하단 그래프] 태그(Tags) 시각화
sns.barplot(data=top_tags, x='Count', y='Name', ax=axes[1], palette='magma')
axes[1].set_title('Top 15 영화 태그 (Tags) 빈도수', fontsize=16, fontweight='bold')
axes[1].set_xlabel('등장 횟수', fontsize=12)
axes[1].set_ylabel('태그', fontsize=12)

for container in axes[0].containers:
    axes[0].bar_label(container, padding=5, fontsize=11, color='black')

for container in axes[1].containers:
    axes[1].bar_label(container, padding=5, fontsize=11, color='black')

# 그래프 간격 조절 및 출력
plt.tight_layout()
plt.show()
```
